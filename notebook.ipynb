```python
#Question 2.

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

#1
df = pd.read_csv('data/land_mines.csv')
print(df.info(), "\n")
print(df.describe(), "\n")
print(df.head(), "\n")
print(df["mine_type"].value_counts(), "\n")
#338 entries, 5 types of mines, all values between 0-1.
#bc all between 0-1 i guess we dont need MinMaxScaler 

#2
#X is what we're looking at, Y is how we identify
X = df[["voltage", "height", "soil"]]
Y = df["mine_type"]
#split 50/50 train
X_train, X_valid, Y_train, Y_valid = train_test_split(
    X,
    Y,
    test_size=0.5,
    random_state=65,
)

#3
#KNN classifier, same structure as regression but accuracy instead of SSE, then we choose the k with highest accuracy
ks = np.arange(1, 51)
best = None
best_accuracy = 0

for k in ks:
    candidate = KNeighborsClassifier(n_neighbors=int(k))
    candidate.fit(X_train, Y_train)

    predictions = candidate.predict(X_valid)
    accuracy = accuracy_score(Y_valid, predictions)

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best = k
print("Best k -", best, "\n")

model = KNeighborsClassifier(n_neighbors=best)
model.fit(X_train, Y_train)
predictions = model.predict(X_valid)

#4
#Create confusion matrice
confusion = pd.crosstab(
    Y_valid,
    predictions,
    rownames=["Actual"],
    colnames=["Predicted"]
)
print(confusion)
print("Accuracy - ", accuracy_score(Y_valid, predictions))
#Accuracy of a 43.78%
#The most accurate mines would be type 2, where we got 30 out of 38 of them correct.

#5
#Though we were highly accurate on type 2 mines, we are still below 50% accuracy in total meaning that every other mine type is very innacurate. I would not recommend someone to base their decision on this model, instead use it as a extra affirmation in their own seperate decision from other information. We don't have enough confidence or information to make a decision off of this model alone, maybe due to sample size or lack of information, regardless it is not sufficient.
```

```python
# Your Phase 2 solution to the problem goes here.
# 1. Regression refers to a numerical outcome prediction, while classsification is used for category predictions.
# 2. Confusion tables are used to compare the predicted vs actual class. It helps us understand model's performance by showing what type of mistakes it could make.
# 3. SSE measures the sum of the squared differences between the predicted and actual values, a lower SSE means the predicted values are closer to the actual values.
# 4. Overfitting is when the model fits the training data too closely, often with a small k, making noise or outliers have too much influence. Underfitting happens when k is too large, making predictions too similar and unable to capture differences.
# 5. Choosing k through the test split instead of the training set is better because it doesn't become specific to that training set. If we chose a k through the entirety of the training set we would find the best k for that specific set, but not the k if we were to add or remove more data. We want a k that performs well on new data, not just the training set.
# 6. 6. A class label gives one clear prediction, the most common class among k neighbors. Class proportions give more information about how the neighbors support each class. One is simpler than the other, but the other would take more time to analyze.
```

```python
#Question 2.

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

#1
df = pd.read_csv('data/land_mines.csv')
print(df.info(), "\n")
print(df.describe(), "\n")
print(df.head(), "\n")
print(df["mine_type"].value_counts(), "\n")

#338 entries, 5 types of mines.
#The five mine types are fairly balanced, with a similar number of observations in each class.
#The feature values for voltage, height, and soil are all between about 0 and 1.
#Since the features are already on similar scales, we don't need MinMaxScaler.

#2
#X is what we're looking at, Y is how we identify
X = df[["voltage", "height", "soil"]]
Y = df["mine_type"]

#split 50/50 train
X_train, X_valid, Y_train, Y_valid = train_test_split(
    X,
    Y,
    test_size=0.5,
    random_state=65,
    stratify=Y,
)

#3
#KNN classifier, same structure as regression but accuracy instead of SSE, then we choose the k with highest accuracy
ks = np.arange(1, 51)
best = None
best_accuracy = 0

for k in ks:
    candidate = KNeighborsClassifier(n_neighbors=int(k))
    candidate.fit(X_train, Y_train)

    predictions = candidate.predict(X_valid)
    accuracy = accuracy_score(Y_valid, predictions)

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best = k

print("Best k -", best, "\n")

model = KNeighborsClassifier(n_neighbors=best)
model.fit(X_train, Y_train)
predictions = model.predict(X_valid)

#4
#Create confusion matrice
confusion = pd.crosstab(
    Y_valid,
    predictions,
    rownames=["Actual"],
    colnames=["Predicted"]
)

print(confusion)
print("Accuracy - ", accuracy_score(Y_valid, predictions))

#Accuracy of a 41.42%
#The most accurate mines would be type 2, where we got 30 out of 35 of them correct.

#5
#Though we were highly accurate on type 2 mines, we are still below 50% accuracy in total meaning that every other mine type is very innacurate. I would not recommend someone to base their decision on this model, instead use it as a extra affirmation in their own seperate decision from other information. There are too few samples for an accurate decision to be made.
```
